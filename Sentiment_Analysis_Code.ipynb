```python
import tweepy
import pandas as pd
import matplotlib.pyplot as plt
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Set up Twitter API credentials
consumer_key = '1VlebIWZjAhtNY5mMmTLiOZkW'  # API Key (Consumer Key)
consumer_secret = 'i75W8UEMSmucOmKjvpBW92vP0nnixtecWLJpWaTlAWV4ywN9Ve'  # API Secret (Consumer Secret)
access_token = '1509113869926793223-3tGHL1ujT8lUcTQbpexwzRLAhQLnbZ'  # Access Token
access_token_secret = '0yY8IfDHnAmgCn0AkDDgtu9sXdLN8zDEu4ogVJNUkyZ7I'  # Access Token Secret

# Authenticate to Twitter
auth = tweepy.OAuthHandler(consumer_key, consumer_secret)
auth.set_access_token(access_token, access_token_secret)
api = tweepy.API(auth)

# Function to get tweets based on a keyword
def get_tweets(keyword, count=50):
    tweets = tweepy.Cursor(api.search, q=keyword, lang='en').items(count)
    tweet_data = []

    for tweet in tweets:
        tweet_data.append(tweet.text)

    return tweet_data

# Function to perform sentiment analysis using VADER
def analyze_sentiment(tweets):
    analyzer = SentimentIntensityAnalyzer()
    sentiment_scores = []
    
    for tweet in tweets:
        score = analyzer.polarity_scores(tweet)
        sentiment_scores.append(score)

    return sentiment_scores

# Function to categorize sentiment as positive, neutral, or negative
def categorize_sentiment(sentiment_scores):
    sentiments = []
    for score in sentiment_scores:
        if score['compound'] >= 0.05:
            sentiments.append('Positive')
        elif score['compound'] <= -0.05:
            sentiments.append('Negative')
        else:
            sentiments.append('Neutral')
    
    return sentiments

# Function to visualize the sentiment distribution
def visualize_sentiment(sentiments):
    sentiment_df = pd.Series(sentiments).value_counts()
    sentiment_df.plot(kind='bar', color=['green', 'red', 'gray'])
    plt.title('Sentiment Distribution')
    plt.xlabel('Sentiment')
    plt.ylabel('Count')
    plt.show()

# Main function to run the analysis
def main():
    # Set the keyword for sentiment analysis (you can change this)
    keyword = input("Enter the keyword/hashtag to analyze: covid19 ")
    tweet_count = int(input("Enter the number of tweets to analyze: 50 "))

    # Get tweets
    print(f"Collecting {tweet_count} tweets related to {keyword}...")
    tweets = get_tweets(keyword, tweet_count)

    # Perform sentiment analysis
    print("Analyzing sentiment...")
    sentiment_scores = analyze_sentiment(tweets)
    sentiments = categorize_sentiment(sentiment_scores)

    # Visualize the sentiment distribution
    print("Visualizing the sentiment distribution...")
    visualize_sentiment(sentiments)

# Run the program
if __name__ == '__main__':
    main()
```
